# Learning Cruve and Perplexity Plot

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
csv_file = "report-en2de-48.2M-params.csv"
data = pd.read_csv(csv_file)
data.head()
```

```python
fix, axes = plt.subplots(ncols=2, figsize=(12, 5))
axes = axes.flatten()


loss_columns = {"Train": "train_loss", "Val": "val_loss"}
ppl_columns = {"Train": "train_perplexity", "Val": "val_perplexity"}

x = list(range(len(data)))
for key, val in loss_columns.items():
    axes[0].plot(x, data[val], label=key)
    axes[0].set_xlabel("Epochs")
    axes[0].set_ylabel("Cross-Entropy")
    axes[0].legend()

for key, val in ppl_columns.items():
    axes[1].plot(x, data[val], label=key)
    axes[1].set_yscale("log")
    axes[1].set_xlabel("Epochs")
    axes[1].set_ylabel("Perplexity (log-scale)")
    axes[1].legend()

plt.show()
```
